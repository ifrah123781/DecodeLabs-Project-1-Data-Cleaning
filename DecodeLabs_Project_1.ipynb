STEP # 21: HANDLE MISSING VALUES PROPERLY

```python
# STEP # 21A: view missing rows first

df[df['CouponCode'].isnull()].head(10)
```

```python
# STEP # 21B: replace missing coupon code

df["CouponCode"] = df["CouponCode"].fillna("No Coupon")
```

STEP # 23: CHECK ALL TEXT COLUMNS FOR EXTRA SPACES

```python
text_columns = ["Product", "PaymentMethod", "OrderStatus", "ReferralSource"]

for column in text_columns:
    print(f"\n{column}:")
    print(df[column].astype(str).str.contains(r"^\s|\s$", regex=True).sum())
```

STEP # 24: CHECK MISSING VALUES AGAIN

```python
df.isnull().sum()
```

STEP # 25: VALIDATE TOTALPRICE

```python
df["CalculateTotal"] = df["Quantity"] * df["UnitPrice"]
df[["Quantity", "UnitPrice", "CalculateTotal"]].head(10)
```

STEP # 26: FIND INCORRECT TOTAL PRICES

```python
df["PriceDifference"] = df["TotalPrice"] - df["CalculateTotal"]
df["PriceDifference"].abs().gt(0.01).sum()
```

STEP # 27: CHECK NUMERIC DATA TYPES

```python
numeric_columns = ["Quantity", "UnitPrice","ItemsInCart", "TotalPrice"]
df[numeric_columns].dtypes
```

STEP # 28: CORRECT/ CONFIRM DATE FORMAT

```python
df["Date"] = pd.to_datetime(df["Date"], errors="coerce")
df["Date"].dtype
```

STEP # 29: FINAL DATA CLEANING VALIDATION

```python
print("Total Rows:", len(df))
print("Total Columns:", len(df.columns))
print("Missing Values:", df.isnull( ).sum().sum())
print("Duplicate Rows:", df.duplicated().sum())
print("Duplicate OrderIDs:", df["OrderID"].duplicated().sum())
print("Invalid Dates:", pd.to_datetime(df["Date"], errors="coerce").isna().sum())
print("Incorrect Total Prices:", df["PriceDifference"].abs().gt(0.01).sum())
```

STEP # 30: REMOVE TEMPORARY COLUMNS

```python
df = df.drop(columns=["CalculateTotal", "PriceDifference"])
df.shape
```
